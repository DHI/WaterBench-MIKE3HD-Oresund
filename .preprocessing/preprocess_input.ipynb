# Preprocessing of input data

The inputs and outputs of the MIKE3-HD model are split into years to make the simulation sequential. For many purposes, however, it is useful to have a single file for each of the inputs and outputs. This notebook concatenates the input data along the time-axis. 

```python
import mikeio.generic
import mikeio
from pathlib import Path
```

```python
fldr = Path(r'\\DKCPH1-STOR.DHI.DK\Projects\11561006\WaterBench\WaterBench-MIKE3HD-Oresund\input')

fldr_out = Path('../input')
```

```python
yr_start = 2023
yr_end = 2025
```

## Boundary conditions

#### North boundary UVTS

```python
file_list = [fldr / "bnd" / f"North_BND_UVTS_{year}.dfsu" for year in range(yr_start, yr_end+1)]
file_list
```

```python
mikeio.generic.concat(file_list, fldr / "North_BND_UVTS.dfsu", keep="last")
```

```python
# Check if it is hourly data
mikeio.open(fldr / "North_BND_UVTS.dfsu").time.freq == 'h'
```

#### South boundary UVTS

```python
file_list = [fldr / "bnd" / f"South_BND_UVTS_{year}.dfsu" for year in range(yr_start, yr_end+1)]
file_list
```

```python
mikeio.generic.concat(file_list, fldr / "South_BND_UVTS.dfsu", keep="last")
```

```python
# Check if it is hourly data
mikeio.open(fldr / "South_BND_UVTS.dfsu").time.freq == 'h'
```

#### North boundary WL

```python
file_list = [fldr / "bnd" / f"North_BND_WL_{year}.dfs1" for year in range(yr_start, yr_end+1)]
file_list
```

```python
mikeio.generic.concat(file_list, fldr / "North_BND_WL.dfs1", keep="last")
```

```python
# Extract only hourly data
print("Frequency is: ", mikeio.open(fldr / "North_BND_WL.dfs1").time.freq)
mikeio.generic.extract(fldr / "North_BND_WL.dfs1", fldr_out / "North_BND_WL.dfs1", step=2)
print("New frequency is: ", mikeio.open(fldr_out / "North_BND_WL.dfs1").time.freq)
```

#### South boundary WL

```python
file_list = [fldr / "bnd" / f"South_BND_WL_{year}.dfs1" for year in range(yr_start, yr_end+1)]
file_list
```

```python
mikeio.generic.concat(file_list, fldr / "South_BND_WL.dfs1", keep="last")
```

```python
# Extract only hourly data
print("Frequency is: ", mikeio.open(fldr / "South_BND_WL.dfs1").time.freq)
mikeio.generic.extract(fldr / "South_BND_WL.dfs1", fldr_out / "South_BND_WL.dfs1", step=2)
print("New frequency is: ", mikeio.open(fldr_out / "South_BND_WL.dfs1").time.freq)
```
