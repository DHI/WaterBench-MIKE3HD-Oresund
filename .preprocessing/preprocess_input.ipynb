# Preprocessing of input data

The inputs and outputs of the MIKE3-HD model are split into years to make the simulation sequential. For many purposes, however, it is useful to have a single file for each of the inputs and outputs. This notebook concatenates the input data along the time-axis. 

```python
import mikeio.generic
import shutil
import mikeio
from pathlib import Path
```

# For publication
In the waterbench publication we want to only use the years 2023-2025 (due to memory constraints on Zenodo).

```python
fldr = Path(r'\\DKCPH1-STOR.DHI.DK\Projects\11561006\WaterBench\WaterBench-MIKE3HD-Oresund\input')

fldr_out = Path('../input')
```

```python
yr_start = 2023
yr_end = 2025
```

## Boundary conditions 

#### North boundary UVTS

```python
file_list = [fldr / "bnd" / f"North_BND_UVTS_{year}.dfsu" for year in range(yr_start, yr_end+1)]
file_list
```

```python
mikeio.generic.concat(file_list, fldr / "North_BND_UVTS.dfsu", keep="last")
```

```python
# Check if it is hourly data
mikeio.open(fldr / "North_BND_UVTS.dfsu").time.freq == 'h'
```

```python
# Copy the file into the input folder here
shutil.copy2(fldr / "North_BND_UVTS.dfsu", fldr_out / "North_BND_UVTS.dfsu")
```
