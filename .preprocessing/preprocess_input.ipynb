# Preprocessing of input data

The inputs and outputs of the MIKE3-HD model are split into years to make the simulation sequential. For many purposes, however, it is useful to have a single file for each of the inputs and outputs. This notebook concatenates the input data along the time-axis. 

```python
import mikeio.generic
from pathlib import Path
```

```python
fldr = Path(r'\\DKCPH1-STOR.DHI.DK\Projects\11561006\WaterBench\WaterBench-MIKE3HD-Oresund\input')
```

## Boundary conditions

#### North boundary UVTS

```python
file_list = [fldr / "bnd" / f"North_BND_UVTS_{year}.dfsu" for year in range(2021, 2026)]
file_list
```

```python
mikeio.generic.concat(file_list, fldr / "North_BND_UVTS.dfsu", keep="last")
```

#### South boundary UVTS

```python
file_list = [fldr / "bnd" / f"South_BND_UVTS_{year}.dfsu" for year in range(2021, 2026)]
file_list
```

```python
mikeio.generic.concat(file_list, fldr / "South_BND_UVTS.dfsu", keep="last")
```

#### North boundary WL

```python
file_list = [fldr / "bnd" / f"North_BND_WL_{year}.dfs1" for year in range(2021, 2026)]
file_list
```

```python
mikeio.generic.concat(file_list, fldr / "North_BND_WL.dfs1", keep="last")
```
