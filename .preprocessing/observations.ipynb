## Prepocessing of observations

Convert from longitude/latitude to Easting/Northing

```python
import pandas as pd
from pyproj import Transformer
```

```python
# --- Configuration ---
INPUT_FILE  = "./stations.csv"   # <-- change to your CSV filename
OUTPUT_FILE = "../observations/stations.csv"
```

```python
# WGS84 (lon/lat degrees) → UTM Zone 33N
# always_xy=True ensures input order is (longitude, latitude)
transformer = Transformer.from_crs("EPSG:4326", "EPSG:32633", always_xy=True)

# --- Load ---
df = pd.read_csv(INPUT_FILE)

# Validate expected columns exist
missing = [c for c in ("Longitude", "Latitude") if c not in df.columns]
if missing:
    raise ValueError(f"Missing columns: {missing}. Found: {list(df.columns)}")

# --- Convert ---
easting, northing = transformer.transform(
    df["Longitude"].to_numpy(),
    df["Latitude"].to_numpy(),
)

df["easting_m"]  = easting.round(1)  
df["northing_m"] = northing.round(1)

# --- Save ---
df.to_csv(OUTPUT_FILE, index=False)
print(f"Done. {len(df):,} rows written to {OUTPUT_FILE}")
print(df.head())
```
