## Convert to dfs1

```python
import pandas as pd
import shutil
from mikeio import ItemInfo, EUMType, EUMUnit

eum = {
    "temperature": (EUMType.Temperature, EUMUnit.degree_Celsius),
    "salinity": (EUMType.Salinity, EUMUnit.PSU),
}

ts_files = sorted(fldr_obs.glob("*_All.csv"))

for csv_file in ts_files:
    df = pd.read_csv(csv_file, index_col="index", parse_dates=True).sort_index()
    # Value column is the one that is not the vertical coordinate
    var = next(c for c in df.columns if c != "z")
    # Long-format profile -> dfs0 with z kept as a separate item.
    # modelskill's VerticalObservation expects the vertical coordinate as an item.
    # A dfs0 allows repeated (equal) timestamps as long as the time axis is sorted.
    ds = mikeio.from_pandas(
        df[[var, "z"]],
        items={var: ItemInfo(*eum[var]), "z": ItemInfo(EUMType.Undefined)},
    )
    dfs0_file = csv_file.with_suffix(".dfs0")
    ds.to_dfs(dfs0_file)
    print(f"Saved {dfs0_file}  (n={ds.n_timesteps})")
```

```python
dest_dir = fldr_obs_raw / "csv_versions"
dest_dir.mkdir(parents=True, exist_ok=True)

for csv_file in ts_files:
    shutil.move(str(csv_file), str(dest_dir / csv_file.name))
    print(f"Moved {csv_file} -> {dest_dir}")
```
