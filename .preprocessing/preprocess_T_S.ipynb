```python
import mikeio
import mikeio.generic
from pathlib import Path
```

```python
fldr_obs = Path('../observations')
fldr_obs_raw = Path('../.preprocessing/raw_data')
```

### FRB9070

```python
ds = mikeio.read(fldr_obs_raw / "FRB9070_S_All.dfs0")
ds
```

```python
ds.sel(time=slice("2023-01-01", "2025-12-31")).to_dataframe().to_csv(fldr_obs / "FRB9070_S_All.csv")
```

```python
ds = mikeio.read(fldr_obs_raw / "FRB9070_T_All.dfs0")
ds
```
