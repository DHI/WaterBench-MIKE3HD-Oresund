# Preprocessing of output data

The inputs and outputs of the MIKE3-HD model are split into years to make the simulation sequential. For many purposes, however, it is useful to have a single file for each of the inputs and outputs. This notebook concatenates the output data along the time-axis. 

```python
import mikeio.generic
import mikeio
from pathlib import Path
import shutil
```

```python
fldr = Path('../output/')
fldr_sample = Path('../output_sample/')
fldr_datastore = Path(r'\\DKCPH1-STOR.DHI.DK\Projects\11561006\WaterBench\WaterBench-MIKE3HD-Oresund\output')
```

## For waterbench publication
We publish the years 2023-2025 in the WaterBench case

```python
yr_start = 2023
yr_end = 2025
```

```python
ds = mikeio.open(fldr / "2021_3D.dfsu")
ds.time.freq
```

```python
ds = mikeio.open(fldr / "2021_2D.dfsu")
ds.time.freq
```

### Extract only hourly timestep

```python
for yr in range(yr_start, yr_end + 1):
    print("Processing year:", yr)
    mikeio.generic.extract(fldr / f"{yr}_3D.dfsu", fldr / f"{yr}_3D_hrly.dfsu", step=2)
    mikeio.generic.extract(fldr / f"{yr}_2D.dfsu", fldr / f"{yr}_2D_hrly.dfsu", step=2)
```

### Concatenate into one file

```python
mikeio.generic.concat([fldr / f"{yr}_3D_hrly.dfsu" for yr in range(yr_start, yr_end + 1)], fldr / f"3D_{yr_start}_{yr_end}_hrly.dfsu")
mikeio.generic.concat([fldr / f"{yr}_2D_hrly.dfsu" for yr in range(yr_start, yr_end + 1)], fldr / f"2D_{yr_start}_{yr_end}_hrly.dfsu")
```

### Copy concatenated file to datastore


```python
shutil.copy2(fldr / f"3D_{yr_start}_{yr_end}_hrly.dfsu", fldr_datastore / f"processed/3D_{yr_start}_{yr_end}_hrly.dfsu")
shutil.copy2(fldr / f"2D_{yr_start}_{yr_end}_hrly.dfsu", fldr_datastore / f"processed/2D_{yr_start}_{yr_end}_hrly.dfsu")
```
